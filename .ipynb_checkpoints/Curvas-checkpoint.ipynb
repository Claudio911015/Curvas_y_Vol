### Repaso tasas de interes

Cualquiera preferiría recibir un peso hoy a recibir un peso en un año, debido a distintos factores tales como liquidez, riesgo de impago o inflación. Por esta razón si depositas un peso hoy esperarías que dentro de un año te sea devuelto más que un peso. Las tasas de interés definirán cuanto “más” de un peso recibirás al final del año.

Antes de ingresar de lleno al tema de tasas es bueno comentar las convenciones de conteo de
días que son más usadas en mercado. Denotaremos por $f_1$ y $f_2$ dos fechas de las cuales
queremos contabilizar su diferencia en años. A continuación se expresa la diferencia en años
bajo cada convención:





Donde $d_1, m_1, y_1$ son correspondientemente el día, mes y año de $f_1$ y $d_2, m_2, y_2$ el día, mes y año de $f_2$

### Tasa simple
Siendo $r$ la tasa simple, el valor de una inversión $N$ al tiempo $t$ estará dado por:

$$V(T) = N (1+rT)$$

Hay que notar que la tasa simple solo paga intereses sobre el monto inicial y no sobre el
monto acumulado del interés.

Es sencillo notar que para obtener un peso al tiempo $T$ es necesario invertir $\frac{1}{1+rT}$
. Nos referiremos a este factor inverso como factor de descuento.

### Tasas compuestas 

Al contrario del interés simple, este tipo de tasas considera el pago de interés sobre el interés
ya obtenido y la inversión original. Al depositar $N$ durante un año al final del año tendremos $N(1+r)$. Si volvemos a tomar ese capital y lo reinvertimos obtendremos $N(1+r)^2$ al final del segundo año. Repitiendo recursivamente este proceso llegaremos a la fórmula de interés compuesto anual.
La fórmula general de interés compuesto está dada por:

$$V(T) = N ({1+\frac{r}{m}})^ {mT}$$

El tamaño de $m$ puede ser tan grande como queramos, lo que deriva en la pregunta ¿qué sucede cuando la tasa se compone de manera instantánea?
Al hacer tender $m$ a infinito obtenemos la fórmula de interés continuo dada por:

$$V(T) =  N e^{rT}$$

### Bono cupon cero

Definimos a un bono cupón cero con vencimiento en $T$ al instrumento financiero que paga 1 al tiempo $T$. Denotaremos por $P(t, T)$ al precio de un bono cupón cero con vencimiento en $T$ al tiempo $t$.
Es fácil notar que $P(t,t) = 1$  para cualquier $t$, pues es un peso pagado de forma inmediata.
En mercado pueden cotizar las tasas de interés o el precio de los bonos cupón cero, el que cotiza y la tasa es implicada a partir de ellos. Aún más, en mercado solo cotizan un conjunto de bonos cupón cero para distintos vencimientos, a partir de los cuales se deriva toda la curva.



Supongan que tienen un bono cupon cero con vencimiento T y su precio $P$. Cual es la tasa simple implicita por el bono ?
$P = 1 / (1+rT)$
$$r = \left(\frac{1}{P}-1\right) \frac{1}{T}$$

## Interpolacion 

Dado que solo conocemos las tasas de interes - factores de descuento en una cantidad finita de puntos necesitamos interpolar el resto de los descuentos a cualquier vencimiento.

Si conocemos el valor de una funcion en solo dos puntos $(x_1, f(x_1))$ y $(x_2, f(x_2))$ la interpolación lineal aproxima cualquier otro punto $(x, f(x))$ por medio de una linea recta 


Cuando tenemos $n$ puntos simplemente interpolamos por pares entre ellos 




Afortunadamente python tiene esto implementado por medio de la funcion 

`from scipy.interpolate import interp1d`

Ejemplo: Las tasas CETEs lineales efectivas anuales actualmente estan dadas por 

$$Cete_{1m} = 11.28 $$
$$Cete_{3m} = 11.35 $$
$$Cete_{6m} = 11.41 $$
$$Cete_{1y} = 11.48 $$

Interpolando encuentra el precio teorico de un Cete a 9 meses.

Hay varias maneras de interpolar entre nodos. Pero la mas comun y la que estaremos usando de aqui en adelante es en la escala $r T$. Si conocemos factores de descuento a varios tenors $$P(0,T_1),P(0,T_2),..., P(0,T_n)$$
calculamos 
$$[-ln(P(0,T_1)),..., -ln(P(0,T_n))] = ...$$

Note que estas cantidades equivales a $r T $ para una composicion continua 

$$P(0, T) = e^ {-rT}$$

Entonces interpolamos lineal en esta escala

$$[r_1T_1, ... ,r_nT_n ]$$

Extrapolamos dejando flat la tasa continua , es decir 

Si  $t \leq T_1$ entonces $r(t) = r(T_1)$ 

Si $t \geq T_n$ entonces $r(t) = r(T_n)$
$$[r_1T_1, ... ,r_nT_n ]$$
$$t$$
$$x = rt$$
$$P(0,t) = e^ {-x} = e^ {-rt}$$

###### Escala tasa

$$[r_1, ... ,r_n ]$$
$$x$$
$$P(0,t) = e^{-x t}$$

```python
from scipy.interpolate import interp1d
import numpy as np

def factor_descuento_cete(t, ts, rs):
    """
    t -> Tiempo al que queremos interpolar
    ts-> Tiempos de los tenors vector
    rs-> Tasas vector
    """
    ps = [1. / (1. + (r * t)) for (r,t) in zip(rs, ts)]
    rts= [-np.log(p) for p in ps]
    if (t >= ts[-1]):
        r =  rts[-1] / ts[-1]
    elif (t <= ts[0]):
        r = rts[0] / ts[0]
    else:
        rateTimeFunctor = interp1d(ts, rts)
        r = rateTimeFunctor(t) / t
    p = np.exp( - r * t)
    return p
print('Descuento a 9 meses es ' +  str(factor_descuento_cete(1/12, ts, rs)))
```

## Tasas Swaps 
Como debieron ver en el modulo de swaps un swap es un conjunto de FRAs. El FRA es un instrumento financiero en que se intercambian dos tasas (una fija y una flotante). El FRA tiene dos fechas relevantes $T$ y $S$, donde $T$ es la fecha en que se fija el valor de la tasa flotante (a.k.a LIBOR, TIIE, etc) y $S$ es la fecha en la que se realiza el pago.

En $S$ un pago fijo calculado a una tasa $K$ es intercambiado contra un pago flotante calculado a la tasa spot $L(S,T)$.

El payoff de este derivado es por lo tanto:







Aunque en un principio podemos reducir la formula de la manera descrita arriba es importante no olvidar la primera formula ya que sera relevante en el futuro


$$(\frac{P(t,T)}{P(t,S)} -1 ) / \tau$$

$$(\frac{1}{P} - 1) / \tau$$

$$1/P = \frac{P(t,T)}{P(t,S)}$$
$$P = \frac{P(t,S)}{P(t,T)}$$

### Mercado mexicano

En el mercado mexicano los swaps estandar pagan un cupon cada 28 dias y se asume que el anio esta compuesto de 360 dias ( 13 periodos de 28 dias). De esta manera podemos asumir para simplificar que 
$$\tau(T_{i-1}, T_i) = \frac{28}{360}$$

Para toda $i$. 

EL vencimiento de los swaps suele darse en terminos de la cantidad de estos pagos en los periodos de 28 dias que el swap tiene. Por ejemplo un 9x1 es un swap con 9 pagos , un 26x1 es un swap con 26 pagos, etc.


```python
from scipy.interpolate import CubicSpline

class Interpolator:
    def interpolate(ts,ps):
        pass

class LogLinearInterpolator(Interpolator):
    def interpolate(self,ts,ps,t):
        #ps: factor de descuento
        #ts : tenors
        rts= [-np.log(p) for p in ps]
        interpolator = interp1d(ts, rts)
        return interpolator(t)/t
    
class LinearInterpolator(Interpolator):
    #Interpolación lineal sobre tasas simples
    def interpolate(self,ts,ps,t):
        rs = [-np.log(p) / t_   for (t_,p) in zip(ts, ps)]
        interpolator = interp1d(ts, rs)
        return interpolator(t)
    
class SplineInterpolator(Interpolator):
    #Interpolación spline sobre tasas simples
    def interpolate(self,ts,ps,t):
        rs = [-np.log(p) / t_   for (t_,p) in zip(ts, ps)]
        interpolator = CubicSpline(ts, rs)
        return interpolator(t)
    
class LogLinearSplineInterpolator(Interpolator):
    #Interpolación spline sobre tasas simples
    def interpolate(self,ts,ps,t):
        if t > 5:
            rs = [-np.log(p) / t_   for (t_,p) in zip(ts, ps)]
            interpolator = CubicSpline(ts, rs)
            return interpolator(t)
        else:
            rts= [-np.log(p) for p in ps]
            interpolator = interp1d(ts, rts)
            return interpolator(t)/t
```

```python
class Curva: 
    # ts: lista con tenors/vencimientos asociados a los factores de descuento
    # ps: lista con factores de descuento
    def __init__(self,obj, ts, ps):
        self._ts = ts 
        self._ps = ps 
        self._interp = obj
        
    def get_ts(self):
        return self._ts
    
    def get_ps(self):
        return self._ps
    
    def descuento(self, t):
        rs= [-np.log(p)/t_ for (p,t_) in zip(self._ps,self._ts)]
        if (t >= self._ts[-1]):
            r =  rs[-1]
        elif (t <= self._ts[0]):
            r = rs[0]
        else:
            r = self._interp.interpolate(self._ts,self._ps,t)
        p = np.exp( - r * t)
        return p
    
    def tasa_forward(self, t1, t2):
        p1 = self.descuento(t1)
        p2 = self.descuento(t2)
        forward = ((p1 / p2) - 1.) / (t2 -  t1)
        return forward
    
    def agregar_nodo(self, t, p):
        self._ts.append(t)
        self._ps.append(p)
        
    def cambiar_ultimo_nodo(self, p):
        self._ps[-1] = p
    
```

```python
ts = [1. / 12., 3. / 12., 6./12., 1.]
rs = [.1128, .1135, .1141, .1148]
ps = [1. / (1. + (r * t)) for (r,t) in zip(rs, ts)]
curvaCetes = Curva(SplineInterpolator(),ts, ps)
print(curvaCetes.tasa_forward(3/12, 1.))
print(curvaCetes.descuento(3/12))
```

#### Valuacion del swap


```python
def fwds(curva, numeroDePagos):
    taus = [28 / 360] * numeroDePagos
    tFinal = np.arange(1, numeroDePagos + 1)
    tFinal = tFinal * 28 / 365
    tInicial = np.arange(0, numeroDePagos)
    tInicial = tInicial * 28 / 365
    forwards = [curva.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)]
    return [tInicial,tFinal,taus,forwards]

def precio_swap(curva, numeroDePagos, nocional, strike):
    #Esta en pesos
    taus = np.array([28 / 360] * numeroDePagos)
    tFinal = np.arange(1, numeroDePagos + 1) * 28 / 365
    tInicial = np.arange(0, numeroDePagos) * 28 / 365
    forwards = np.array([curva.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)])
    descuentos = np.array([curva.descuento(t) for t in tFinal])
    swap = nocional * np.sum(taus * descuentos * (forwards - strike ))
    return swap

def tasa_swap(curva, numeroDePagos):
    #Esta en orcentaje
    taus = np.array([28 / 360] * numeroDePagos)
    tFinal = np.arange(1, numeroDePagos + 1) * 28 / 365
    tInicial = np.arange(0, numeroDePagos) * 28 / 365
    forwards = np.array([curva.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)])
    descuentos = np.array([curva.descuento(t) for t in tFinal])
    numerador = np.sum(descuentos * taus * forwards )
    denominador = np.sum(descuentos * taus)
    return numerador / denominador
```

### Bootstrapping 


Este proceso recursivo se continúa hasta obtener los $n$ factores $P_1^*, ..., P^*_n$.

Note que un problema potencial de este metodo es que solo conocemos las parswaps a algunos vencimientos , por lo cual no podremos despejar cada uno de los factores de descuento , por lo cual tendremos que recurrir a interpolacion para implicar valores intermedios.


Repetimos este proceso hasta que conozcamos todos los descuentos a los vencimientos de los swaps

## Calibracion de curva


```python
from scipy import optimize
def funcion_swap_a_optimizar(tasa, curva, s, numeroDePagos):
    """
    s: Tasa swap
    tasa: tasa composicion continua
    t: tiempo para calcular el descuento con tasa
    """
    t = numeroDePagos * 28. / 365.
    p = np.exp(-t * tasa)
    curva.cambiar_ultimo_nodo(p)
    swapCurva = tasa_swap(curva, numeroDePagos)    
    return 10*(swapCurva - s) 
    
    
def calibrar_curva_swaps(par_swaps, tenors,interp, log=False):
    """
    par_swaps: Vector con tasas par swap 
    tenors: enteros, por ejemplo si tenemos swaps [3x1, 9x1, 13x1] el input seria [3, 9, 13]
    interp: Método de interpolación (Linear, LogLinear, Spline)
    """
    t = np.array(tenors) * 28 / 365
    p = np.exp(-np.array(par_swaps) * t)
    curva = Curva(interp,[],[])
    for i in range(0, len(tenors)):
        # Anadir un nuevo punto a la curva que corresponda al vencimiento del swap 
        curva.agregar_nodo(t[i], p[i])
        tasa = optimize.newton(funcion_swap_a_optimizar, par_swaps[i], 
                               args =(curva, par_swaps[i], tenors[i]))
        if log:
            print(" Swap " + str(tenors[i]) + 'x1 calibrated ' + str(tasa)) 
    return curva   
```

### Calendarios 

Estamos trabajando en un mundo simplificado donde no estamos considerando fechas , si no tiempos. Pero vale la pena especificar que en el mundo real los tiempos de pago y accrued son determinados por fechas y las fechas solamente pueden estar en dias habiles. Si alguna fecha cae en un dia inhabil , la fecha del calendario se "patea" al siguiente dia habil.

Las vacaciones que usamos para determinar depende del instrumento. Usualmente si solo una moneda esta involucrada  (por ejemplo un swap en pesos mexicanos) usamos solo las vacaciones del pais de esa moneda (por ejemplo las vacaciones en Mexico). 

Si dos o mas monedas estan involucradas ( por ejemplo un forward peso-dolar) usamos el calendario de ambos paises (vacaciones de Mexico y Estados Unidos)

```python
from datetime import date
import datetime
def es_dia_habil(fecha, holidays):
    if (fecha in holidays) or (fecha.weekday() > 4):
        return False
    else:
        return True

holidays =  [date(2023, 9, 16), date(2023, 11, 2), date(2023, 12, 12), date(2024, 2, 6), date(2024, 3, 20), 
             date(2024, 4, 6), date(2024, 4, 7), date(2024, 5, 1), date(2024, 1, 16)]

calendario = [date(2023, 8, 1)]
for i in range (0, 26):
    fechaSiguiente = calendario[i] + datetime.timedelta(days=28)
    while not  es_dia_habil(fechaSiguiente, holidays):
        fechaSiguiente = fechaSiguiente + datetime.timedelta(days=1)
    calendario.append(fechaSiguiente)
calendarioTs = calendario[1:]
```

### FX Forwards

Un forward en FX es un contrato que te da el derecho y la obligacion de comprar 1 unidad de la divisa foranea a cambio de un precio prestablecido ($K$) de la divisa local, en un tiempo futuro $T$.

Por que nos interesan estos intrumentos al construir curvas? Por su valuacion 

Sin perdida de generalidad, vamos a suponer que las dos monedas con las que trabajaremos son $𝑐𝑐𝑦_1=𝑀𝑋𝑁$ y $𝑐𝑐𝑦_2=𝑈𝑆𝐷$, pero lo discutido a continuación es aplicable a cualquier otro par de monedas

Forward: Da el derecho y la obligación de comprar 1 dólar USD a un precio de 𝐾 pesos MXN  en una fecha futura 𝑇.

$𝑆_𝑇$ es el tipo de cambio peso/dólar (USD/MXN) en la fecha de vencimiento 𝑇

$$(S_T - K)$$





Podemos replicar el payoff del forward con el siguiente portafolio




 Queremos encontrar $K = 𝐹(𝑡, 𝑇)$  tal que  $𝑆_𝑡 𝑃^{𝑈𝑆𝐷} (𝑡, 𝑇) -K𝑃^{𝑀𝑋𝑁} (𝑡,𝑇) =0$

El strike que hace 0 este precio es:
$𝐹(𝑡, 𝑇)=𝑆_𝑡   \frac{𝑃^{𝑈𝑆𝐷} (𝑡, 𝑇)}{ 𝑃^{𝑀𝑋𝑁} (𝑡,𝑇)}  $

De la ecuacion anterior conocemos $S_t$ (es el tipo de cambio al dia de hoy. Tambien $F(t,T)$ se cotiza en el mercado para ciertos tenors (describiremos a detalle abajo), y suponiendo que la curva $USD$ se calibra de forma independiente entonces conocemos tambien $𝑃^{𝑈𝑆𝐷} (𝑡, 𝑇)$.

De esta manera la unica incognita en esta ecuacion es $𝑃^{𝑀𝑋𝑁} (𝑡,𝑇)$, por lo cual podriamos implicitar una curva en el short end por medio de las cotizaciones de forwards. Despejando 

$$P^{MXN}(t, T) = \frac{S_t 𝑃^{𝑈𝑆𝐷} (𝑡, 𝑇)}{𝐹(𝑡, 𝑇)} $$


En el el mercado se cotizan los precios forwards ( de manera indirecta)  a traves de los forward points para tenors cortos (usualmente menores a 1 año)

La definicion de forward point en el tenor $T$ esta dado por 

$FwdPoint_T = 10,000   (F(t,T) - S_t)$

Es decir, es la diferencia entre el precio FX forward y el precio FX spot en puntos base.

De esta manera es facil despejar 

$$F(t,T) = S_t + \frac{FwdPoint_T }{10000}$$






### Ejemplo:
Calibra una curva MXN con la informacion de los forwards de mercado siguiente



La curva de dolares fue calibrada de forma independiente y puede ser econtrada abajo. El tipo de cambio USDMXN es de 17

```python
def calibrar_curva_FX(tenors, fxPoints, fxSpot, curvaUSD):
    psMXN = []
    for (tenor, fxPoint) in zip(tenors, fxPoints):
        F = fxSpot + fxPoint/10000.
        p = (fxSpot / F) * curvaUSD.descuento(tenor)
        psMXN.append(p)
    curvaMXN = Curva(LinearInterpolator(),tenors, psMXN)
    return curvaMXN
        
```

### Tasas de referencia (RFR)

Durante mucho tiempo las tasas de estilo  IBOR (tasas de composicion simple a tenors de 1M, 3M, 6M, etc) fueron las principales en el mercado, usadas para la determinacion del precio de multiples instrumentos derivados . Sin embargo esa epoca ha terminado para los mercados mas grandes del mundo ( como USA y Europa) buscando hacer una transicion hacia las conocidas "Tasas de referencia" .

Buena parte del empuje hacia el cambio obedece a la necesidad de reforzar la integridad del mercado tras los casos de conducta inapropiada relacionados con las tasas comunicadas por los bancos para el
cálculo del LIBOR. Hay casos muy sonados de como grandes bancos manipularon las tasas de interes LIBOR para beneficio propio. https://en.wikipedia.org/wiki/Libor_scandal

Desde el punto de vista técnico, el panel de bancos que participaba con sus datos en la elaboración del LIBOR se redujo durante la crisis. A partir de entonces no solo hay menos bancos (ahora son entre 11 y 16 bancos), sino que hay muchas menos transacciones y, por tanto, ha perdido representatividad: el resultado ya no reflejaba la realidad del mercado


Inicialmente, la atención se ha centrado en la creación de RFR a *un día*  basadas en transacciones en mercados monetarios suficientemente líquidos. Los mercados al contado y de derivados que utilizan las nuevas RFR como referencia ya son muy operados y liquidos.

#### Caracteristicas deseables

i. Ser una representación fiel y sólida de las tasas de interés en los principales
mercados monetarios, a prueba de manipulación. 

ii. Servir como tasa de referencia para contratos financieros más allá del mercado monetario. Así, la  asa debería poder utilizarse para descontar y determinar el precio de instrumentos al contado y de derivados sobre tasas de interés. Por ejemplo, los contratos de swaps sobre índices a un día (OIS) de
distintos vencimientos deberían poder utilizar fácilmente esta tasa como referencia, generando una curva OIS para determinar el precio de contratos a plazos más largos

iii. Servir como tasa de referencia para los préstamos y la financiación a plazo.

#### Riesgo de credito 

Otra razon por la cual se realiza la migracion es por el riesgo de credito. Despues de la crisis del 2008 en la que el riesgo de credito tomo relevancia se ha decidido que es una mejor idea tener prestamos a mucho mas corto plazo. 

Las tasas LIBOR eran una referencia de prestamos interbancarios a cierto plazo. Por ejemplo la LIBOR mas popular era a 3 meses , por lo cual representaba un prestamo a 3 meses. Si nos imaginamos que un banco cae en default al inicio del prestamo hay 3 meses de incertidumbre si nos pagaraeste prestamo o no. 

Adicionalmetne estamos dejando tres meses enteros en los que la contraparte puede caer en default. Por eso se adopto una aproximacion que mitigue el riesgo de credito , haciendo prestamos a un horizonte de un dia.



### SOFR

Es una tasa de interés que mide el costo del dinero en dólares americanos, en un plazo de un día y que es calculada con base en operaciones de crédito garantizadas.

* Es una tasa de un día.

* Nace de operaciones reales pactadas en el mercado.

* "Es libre de riesgo", dado que las operaciones utilizadas para calcular el índice están siendo garantizadas con títulos emitidos por el Tesoro de los Estados Unidos, una garantía de alta calidad crediticia.








### TIIE de FONDEO

En Mexico se trata de hacer una tasa de referencia a 1 dia a traves de la TIIE de FONDEO. En general la tasa de TIIE es una tasa de fondeo interbancario. El tenor de TIIE "usual" es el de 28 dias ( que hemos discutido antes). La diferencia con TIIE de FONDEO es que el tenor es de 1 dia habil.

La TIIE de Fondeo se basa en las operaciones de fondeo al mayoreo en pesos realizadas por la banca y casas de bolsa a través de operaciones de reporto a plazo de un día hábil bancario con títulos emitidos por el Gobierno Federal, el Instituto de Protección al Ahorro Bancario (IPAB) y el Banco de México. La TIIE de Fondeo se determina como la mediana ponderada por volumen de la tasa de interés pagada en las transacciones de fondeo mencionadas en anteriormente.

Nota: Se conoce como Reporto a las operaciones en las que una de las partes (el reportador) compra a la otra (reportado, quien vende) títulos valores públicos o privados, contado, y simultáneamente convienen la operación inversa de compra/venta ("recompra") a un plazo determinado, por un precio convenido que se denomina prima. Es decir hoy vendo bonos con la promesa de comprarlos de nuevo mañana (o saco un prestamo dejandote mis bonos como colateral).

Podemos ver que similarmente a SOFR pretende ser una tasa a 1 dia, libre de riesgo por que esta colateralizada por bonos emitidos por el gobierno mexicano.

## OIS (Overnight Index Swap)

Como son los swaps sobre esta nueva clse de tasas a un dia y como construimos curvas a partir de ellos?

Un OIS (Overnight Index Swap) tiene dos patas, una pata fija que paga una tasa fija y una tasa flotante que paga cupones OIS referenciados a la tasa a un dia.
### Cupones OIS pata fija 

Los cupones de la pata fija del OIS estan dados por 

$$ N K \tau(F_1, F_2) $$

Donde 

$F_1$ es la fecha de inicio del cupon 

$F_2$ fecha final del cupon 

$\tau(F_1, F_2) $ fraccion entre las fechas $F_1$ y $F_2$

$N$ nocional

$K$ tasa fija del swap


### Cupones OIS pata flotante

Los cupones de la pata flotante del OIS estan dados por 

$$ N r \tau(F_1, F_2) $$



La tasa $r$ esta determinada como una composicion de las tasas diarias en este periodo. 

Consideremos que las fechas habiles entre $F_1$ y $F_2$ estan dadas por  ($f_0, f_1, ..., f_n$); Para $f_0 = F_1$ y $f_n=F_2$. Y las tasas de composicion diaria en estas fechas son ($r_1,..., r_n$).

De igual manera calculamos ($\tau_1,..., \tau_n$) como la fraccion entre las fechas $f_i$ y $f_{i-1}$ 

$$r = \frac{1}{\tau(F_1, F_2)}\left( \prod_{i=1}^n (1 + \tau_i r_i) - 1 \right)$$


#### Ejemplo: 

Las tasas SOFR del proximo mes (variables `tasas`) estan dadas abajo. Calcula el cupon que se pagara en un OIS entre Julio 23 y Agosto 23, para un nocional de 10,000

```python
tasas = [1.1, 1.2, 1.3, 1.4, 1.5, 1.6, 1.7, 1.8, 1.9, 2.0, 2.1, 
         2.2, 2.3, 2.4, 2.3, 2.4, 2.5, 2.6, 2.7, 2.8 , 
         2.9, 3., 3.1, 3.2, 3., 3.4]
```

Solucion: Primero calculamos los dias habiles entre estas dos fechas 

```python
def fechas_habiles_entre(startDate, endDate, holidays):
    # Regresa todas las fechas habiles entre startDate y endDate
    fechas = []
    fechaSiguiente = startDate
    while fechaSiguiente <= endDate:
        fechas.append(fechaSiguiente)
        fechaSiguiente = fechaSiguiente + datetime.timedelta(days=1)
        while not  es_dia_habil(fechaSiguiente, holidays): 
            fechaSiguiente = fechaSiguiente + datetime.timedelta(days=1)
    return fechas
```

```python
from datetime import date
startDate = date(2023, 7, 19)
endDate = date(2023, 8, 19)
dias = fechas_habiles_entre(startDate, endDate, holidays)
dias
```

Calculemos la tasa cupon

```python
def calculo_cupon_ois(startDate, endDate, nocional, tasas):
    dias = fechas_habiles_entre(startDate, endDate, holidays)
    startDates = dias[:-1]
    endDates = dias[1:]
    taus = [(end - st).days/360 for (st, end) in zip (startDates, endDates)]
    producto = np.product(1 + (taus * np.array(tasas)))
    tau = (endDate - startDate).days /  360
    r = (1/tau) * (producto - 1.)
    cupon = nocional * r  * tau
    return cupon
```

## Mercado OIS SOFR

En Estados Unidos cotizan OIS para la tasa SOFR a varios tenors $1y, 2y, 3y, 5y, 7y 10y, 15y, 30y$. Los swaps cotizados tienen una frequencia anual (es decir el pago de cupon sucede cada año). AL igual que con los swaps normales se cotizan las tasas fijas que hacen que el valor del swap sea 0 al dia de hoy.

Podemos implicar las tasas de descuento a partir de las cotizaciones de estos swap en una forma muy similar en que lo hicmos para los swaps normales.

Si recordamos el proceso, como primer paso encontrabamos el factor de descuento que hiciera que el primer swap valiera 0. Posteriormente encontrabamos el segundo factor de descuento que en combinacion con el factor ya calibrado hiciera que el segundo swap valiera , etc. 

El proceso aqui es exactamente el mismo, aunque la valuacion de los swaps sea ligeramente diferente segumos la  misma metodologia 


Recordando como lo hicimos antes el primer paso es crear una funcion que reciba una curva y nos devuelva el precio del OIS

```python
def start_and_end_dates_ois(startDate, numeroDeAnios):
    fechas = [startDate]
    fechaSiguiente = startDate
    for i in range(1, int(numeroDeAnios) + 1):
        fechaSiguiente = fechaSiguiente + datetime.timedelta(days=365)
        #Patear a siguiente fecha habil
        while not  es_dia_habil(fechaSiguiente, holidays): 
            fechaSiguiente = fechaSiguiente + datetime.timedelta(days=1)
        fechas.append(fechaSiguiente)
    startDates = fechas[:-1]
    endDates = fechas[1:]
    return startDates, endDates
```

```python
fecha1 = date(2022, 7,19)
numeroDeAnios = 10
startDates, endDates = start_and_end_dates_ois(startDate, numeroDeAnios)
taus = [(endDate - startDate).days / 360 for (startDate, endDate) in zip(startDates, endDates)]
taus
```

```python
def tasas_sofr_proyectadas(curva, today, dailyStartDates, dailyEndDates):
    fwds =  []
    for (st, end) in zip(dailyStartDates, dailyEndDates):
        tau = (end - st).days / 360.
        tEnd = (end - today).days / 365.
        tStart = (st - today).days / 365.
        fwd = curva.tasa_forward(tStart, tEnd)
        fwds.append(fwd)
    return fwds
```

```python
def ois_swap(curva, today , startDate, numeroDeAnios, nocional, strike):
    couponStartDates, couponEndDates = start_and_end_dates_ois(startDate, numeroDeAnios)
    taus = [(end - start).days / 360. for (start, end) in zip(couponStartDates, couponEndDates)]
    ts = [(endDate - today).days / 365. for endDate in couponEndDates]
    pv = 0.
    for (couponStart, couponEnd, t, tau) in zip(couponStartDates, couponEndDates, ts, taus):
        dias = fechas_habiles_entre(couponStart, couponEnd, holidays)   
        dailyStartDates = dias[:-1]
        dailyEndDates = dias[1:]
        descT = curva.descuento(t)
        tasas = tasas_sofr_proyectadas(curva, today, dailyStartDates, dailyEndDates)
        cuponFlotante = calculo_cupon_ois(couponStart, couponEnd, nocional, tasas)
        cuponFijo = strike * tau * nocional
        pv = pv + (cuponFlotante - cuponFijo) *  descT
    return pv
```

Primero hagamos una funcion que calcule la tasa "par swap" OIS, ya que como en la calibracion de swap buscaremos que la parswap sea igual a la del mercado 


```python
def ois_parswap(curva, today , numeroDeAnios):
    couponStartDates, couponEndDates = start_and_end_dates_ois(startDate, numeroDeAnios)
    taus = [(end - start).days / 360. for (start, end) in zip(couponStartDates, couponEndDates)]
    ts = [(endDate - today).days / 365. for endDate in couponEndDates]
    pvFloat = 0.
    pvFix = 0.
    for (couponStart, couponEnd, t, tau) in zip(couponStartDates, couponEndDates, ts, taus):
        dias = fechas_habiles_entre(couponStart, couponEnd, holidays)   
        dailyStartDates = dias[:-1]
        dailyEndDates = dias[1:]
        # Proyectemos las tasas diaria 
        descT = curva.descuento(t)
        tasas = tasas_sofr_proyectadas(curva, today, dailyStartDates, dailyEndDates)
        pvFloat= pvFloat + (calculo_cupon_ois(couponStart, couponEnd, 1., tasas) * descT) 
        pvFix = pvFix + (descT * tau)
    return pvFloat / pvFix
```

```python
s = ois_parswap(curvaUSD, today , 10)
s
```

```python
from scipy import optimize
def funcion_ois_a_optimizar(tasa, curva, s, numeroDeAnios, today):
    """
    s: Tasa swap
    tasa: tasa composicion continua
    t: tiempo para calcular el descuento con tasa
    """
    p = np.exp(-numeroDeAnios * tasa)
    curva.cambiar_ultimo_nodo(p)
    return 100000. * (ois_parswap(curva, today , numeroDeAnios) - s)
    
```

```python
def calibrar_curva_ois(par_swaps, tenors,interp, today):
    """
    par_swaps: Vector con tasas par swap ois
    tenors: numero de años en el OIS. Esta variable esta en años
    interp: Método de interpolación a utilizar
    """
    t = np.array(tenors)
    p = np.exp(-np.array(par_swaps) * t)  
    curva = Curva(interp,[], []) 
    for i in range(0, len(tenors)):
        # Anadir un nuevo punto a la curva que corresponda al vencimiento del swap 
        couponStartDates, couponEndDates = start_and_end_dates_ois(today, tenors[i])
        curva.agregar_nodo(t[i], par_swaps[i])
        tasa_i = optimize.newton(funcion_ois_a_optimizar, par_swaps[i],
                                 args =(curva, par_swaps[i], tenors[i], today))
        curva.cambiar_ultimo_nodo(np.exp(-tenors[i] * tasa_i))
        print(" OIS rate " + str(tenors[i]) + 'y calibrated ' + str(tasa_i)+
              "; Parswap "+ str(ois_parswap(curva, today , tenors[i])) ) 
    return curva   
```

### Curvas colateralizadas en otras monedas

Como vimos en la seccion anterior existen instrumentos que por su naturaleza implican una curva colateralizada. Por ejemplo los factores de descuento de  SOFR, estan basados en operaciones de reporto colateralizadas con bonos del tesoro estadunidense . De igual manera si construyeramos una curva de TIIE de FONDEO los factores de descuento implicarian una colateralizacion con bonos del gobierno mexicano.

Sin embargo puede haber inversionistas extranjeros, por ejemplo americanos,  que quieran entrar en algun derivado en pesos, pero que quieran que su colateral este dado en dolares. Como construimos una curva para estos inversionistas? 

#### Cross currency swaps 


Esta clase de instrumento son bastante liquidos y operados con regularidad en el mercado.


Sin embargo podemos ir a ver mercado y darnos cuenta que hay spreads para cross currencies USD vs MXN distintos de 0. Asi que resumiendo , si vamos a mercado podemos entrar a precio de 0 a

    a) Un swap de mercado
    b) Un cross currency con spread distinto de 0 
En nuestro planteamiento actual esto es imposible, por lo cual debemos cambiar nuestro planteamiento. La alternativa que se ha tomado es suponer que dentro de una misma moneda existen dos curvas. Una usada para estimar las tasas futuras y otra usada solamente para descontar 



Recordemos por ejemplo la valuacion de nuestro swap estandar:





Podemos ver que ls factores de descuento "juegan" un rol en dos parte, la primera la estimacion de tasas forward y la segunda los factores de descuento para traer los flujos a valor presente.

Anteriormente usabamos la misma curva de descuento para la parte azul y la roja , pero esta nueva aproximacion nos dice que construyamos una curva azul (curva de estimacion) y una roja (curva de descuento).

Como veremos adelante esto se vuelve un sistema de dos ecuaciones (los datos de mercado de los swaps y los cross currency) con dos incognitas (la curva de estimacion y la curva de descuento).

Asi que procedamos a calibrar dos curvas al mismo tiempo:

### Calibracion de curvas usando cross currency swaps

Para poder construir estas dos curvas necesitamos como es usual hacer un pricer de 

    a) Cross currency swap
    b) Rehacer el pricer de swap para que use dos curvas
    
   

#### Pricing swap con dos curvas

# MXN Collat SOFR , MXN Basis SOFR

```python
def precio_swap2(curvaEst, curvaDesc, numeroDePagos, nocional, strike):
    taus = np.array([28 / 360] * numeroDePagos)
    tFinal = np.arange(1, numeroDePagos + 1) * 28 / 365
    tInicial = np.arange(0, numeroDePagos) * 28 / 365
    forwards = np.array([curvaEst.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)])
    descuentos = np.array([curvaDesc.descuento(t) for t in tFinal])
    swap = nocional * np.sum(taus*descuentos*(forwards-strike))
    return swap

def tasa_swap2(curvaEst, curvaDesc, numeroDePagos):
    taus = np.array([28 / 360] * numeroDePagos)
    tFinal = np.arange(1, numeroDePagos + 1) * 28 / 365
    tInicial = np.arange(0, numeroDePagos) * 28 / 365
    forwards = np.array([curvaEst.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)])
    descuentos = np.array([curvaDesc.descuento(t) for t in tFinal])
    return np.sum(taus*descuentos*forwards) / np.sum(taus*descuentos)
```

#### Pricing cross currency 

```python
def precio_xccy(curvaEstDOM, curvaDescDOM, curvaFOR, numeroDePagos, nocionalDOM, 
                fxToday , spread):
    nocionalFOR = - nocionalDOM / fxToday
    taus = np.array([28 / 360] * numeroDePagos)
    tFinal = np.arange(1, numeroDePagos + 1) * 28 / 365
    tInicial = np.arange(0, numeroDePagos) * 28 / 365
    
    forwardsDOM = np.array([curvaEstDOM.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)])
    descuentosDOM = np.array([curvaDescDOM.descuento(t) for t in tFinal])
    pataDOM =  nocionalDOM - (nocionalDOM * descuentosDOM[-1])
    pataDOM =  pataDOM - np.sum(taus * nocionalDOM * descuentosDOM * forwardsDOM)
    
    forwardsFOR = np.array([curvaFOR.tasa_forward(t1, t2) for (t1, t2) in zip(tInicial, tFinal)])
    descuentosFOR = np.array([curvaFOR.descuento(t) for t in tFinal])
    pataFOR = nocionalFOR -(nocionalFOR * descuentosFOR[-1])
    pataFOR = pataFOR - np.sum((taus * nocionalFOR * descuentosFOR * (forwardsFOR + spread)))
    
    xccy = pataDOM + (pataFOR * fxToday)
    return xccy
```

### Calibrar curva con  colateral dolares

```python
def funcion_swap_xccy_a_optimizar(tasas, curvaEst, curvaDesc, curvaFOR, 
                                  parswap, spread, numeroDePagos, fxToday):
    """
    tasa = vector que tiene tasas de estimación y tasas de descuento
    s: Tasa swap
    tasa: tasa composicion continua
    t: tiempo para calcular el descuento con tasa
    """
    tasaEst, tasaDesc = tasas
    t = numeroDePagos * 28. / 365.
    pEst = np.exp(-t * tasaEst)
    pDesc =  np.exp(-t * tasaDesc)
    curvaEst.cambiar_ultimo_nodo(pEst)
    curvaDesc.cambiar_ultimo_nodo(pDesc)
    nocionalDOM = 100_000_000_000.
    swapPrecio = precio_swap2(curvaEst, curvaDesc, numeroDePagos, nocionalDOM, parswap)
    xccyPrecio = precio_xccy(curvaEst, curvaDesc, curvaFOR, numeroDePagos, nocionalDOM, 
                             fxToday , spread)
    return [swapPrecio *nocionalDOM , xccyPrecio*nocionalDOM]
    
def funcion_swap_xccy_a_optimizar_v2(tasas, curvaEst, curvaDesc, curvaFOR, 
                                  parswap, spread, numeroDePagos, fxToday):
    """
    tasa = vector que tiene tasas de estimación y tasas de descuento
    s: Tasa swap
    tasa: tasa composicion continua
    t: tiempo para calcular el descuento con tasa
    """
    tasaEst, tasaDesc = tasas
    t = numeroDePagos * 28. / 365.
    pEst = np.exp(-t * tasaEst)
    pDesc =  np.exp(-t * tasaDesc)
    curvaEst.cambiar_ultimo_nodo(pEst)
    curvaDesc.cambiar_ultimo_nodo(pDesc)
    nocionalDOM = 100_000_000_000.
    swapPrecio = precio_swap2(curvaEst, curvaDesc, numeroDePagos, nocionalDOM, parswap)
    xccyPrecio = precio_xccy(curvaEst, curvaDesc, curvaFOR, numeroDePagos, nocionalDOM, 
                             fxToday , spread)
    return (swapPrecio**2) *nocionalDOM + (xccyPrecio**2)*nocionalDOM
    
    
def calibrar_curvas_swap_xccy(par_swaps, xccys_spreads, tenors, curvaFOR, fxToday,
                                  interp_desc,interp_estim):
    """
    par_swaps: Vector con tasas par swap 
    tenors: enteros, por ejemplo si tenemos swaps [3x1, 9x1, 13x1] el input seria [3, 9, 13]
    xccys_spreads Spreads de cross currency swaps
    """
    t = np.array(tenors) * 28 / 365
    p = np.exp(-np.array(par_swaps) * t)   
    curvaEst = Curva(interp_estim, [], [])
    curvaDesc = Curva(interp_desc, [], [])
    for i in range(0, len(tenors)):
        # Anadir un nuevo punto a la curva que corresponda al vencimiento del swap
        curvaEst.agregar_nodo(t[i], p[i])
        curvaDesc.agregar_nodo(t[i], p[i])
        # Encuentra en donde alcanza el minimo de la funcion  
        tasas = optimize.newton(funcion_swap_xccy_a_optimizar,
                       x0 = [par_swaps[i], par_swaps[i]],
                       args = (curvaEst, curvaDesc, curvaFOR, par_swaps[i], 
                               xccys_spreads[i], tenors[i], fxToday))
        tasaEst = tasas[0]   
        tasaDesc = tasas[1]
        curvaEst.cambiar_ultimo_nodo(np.exp(-t[i] * tasaEst))
        curvaDesc.cambiar_ultimo_nodo(np.exp(-t[i] * tasaDesc))
        
        print(" Curvas nodo " + str(tenors[i]) + 'x1 calibrated '  ) 
    return curvaEst, curvaDesc   
```

De manera sintética, podemos replicar un forward FX operando en el mercado de divisas al contado y realizando simultáneamente un préstamo y un depósito para igualar la posición al contado y replicar así un tipo de cambio a plazo FX. 

Matemáticamente, podemos expresar esto de la siguiente manera:



```python
TIIE = curvaEst
MXN_BASIS_USD = curvaDesc
USD_BASIS_SOFR = curvaFOR
```
