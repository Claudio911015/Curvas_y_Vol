```python
import numpy as np
import matplotlib.pyplot as plt

# Función de ejemplo
def f(x):
    return x**3 - 2*x**2 + 4*x - 8

# Derivada de la función de ejemplo
def f_prime(x):
    return 3*x**2 - 4*x + 4



# Método de Bisección
def bisection_method(func, a, b, tol=1e-6, max_iter=100):
    iter_count = 0
    while (b - a) / 2 > tol and iter_count < max_iter:
        c = (a + b) / 2
        if func(c) == 0:
            return c
        elif func(a) * func(c) < 0:
            b = c
        else:
            a = c
        iter_count += 1
    return (a + b) / 2

# Método de Newton-Raphson
def newton_raphson(func, func_prime, x0, tol=1e-6, max_iter=100):
    x = x0
    iter_count = 0
    while abs(func(x)) > tol and iter_count < max_iter:
        x = x - func(x) / func_prime(x)
        iter_count += 1
    return x

# Intervalo inicial para bisección
a_bisection = 0.2
b_bisection = 2.6

# Valor inicial para Newton-Raphson
x0_newton = 25

# Encontrar las raíces usando ambos métodos
root_bisection = bisection_method(f, a_bisection, b_bisection, max_iter=50)
root_newton = newton_raphson(f, f_prime, x0_newton, max_iter=50)

# Crear un rango de valores para la gráfica
x_vals = np.linspace(0, 4, 400)
y_vals = f(x_vals)

# Graficar la función y las raíces encontradas
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label='f(x)')
plt.scatter([root_newton], [f(root_newton)], color='red', label='Roots Newton')
plt.scatter([root_bisection], [f(root_bisection)], color='blue', label='Roots Bisect')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Método de Bisección y Newton-Raphson')
plt.legend()
plt.grid()
plt.show()
```
